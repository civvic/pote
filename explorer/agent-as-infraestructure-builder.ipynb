## A local Python weather tool

This version uses:

- **Open-Meteo geocoding API** to resolve a place name to coordinates
- **Open-Meteo forecast API** to get current conditions and today’s min/max

I chose it because the official docs indicate:

- there is a geocoding endpoint,
- there is a forecast endpoint,
- and for non-commercial use no API key is required.

I also made it:

- **stdlib-only**: no third-party packages
- **cache-aware**: geocoding is cached locally, so repeated runs for the same city only need the forecast request
- **fast and cheap**: after the first run, it is basically one HTTP GET plus formatting

```python
#!/usr/bin/env python3
import json
import sys
import urllib.parse
import urllib.request
from pathlib import Path

CACHE_DIR = Path.home() / ".cache" / "simple_weather"
CACHE_DIR.mkdir(parents=True, exist_ok=True)
GEOCODE_CACHE = CACHE_DIR / "geocode.json"

WEATHER_CODES = {
    0: "clear sky",
    1: "mainly clear",
    2: "partly cloudy",
    3: "overcast",
    45: "fog",
    48: "depositing rime fog",
    51: "light drizzle",
    53: "moderate drizzle",
    55: "dense drizzle",
    56: "light freezing drizzle",
    57: "dense freezing drizzle",
    61: "slight rain",
    63: "moderate rain",
    65: "heavy rain",
    66: "light freezing rain",
    67: "heavy freezing rain",
    71: "slight snow",
    73: "moderate snow",
    75: "heavy snow",
    77: "snow grains",
    80: "slight rain showers",
    81: "moderate rain showers",
    82: "violent rain showers",
    85: "slight snow showers",
    86: "heavy snow showers",
    95: "thunderstorm",
    96: "thunderstorm with slight hail",
    99: "thunderstorm with heavy hail",
}

def fetch_json(url: str) -> dict:
    req = urllib.request.Request(
        url,
        headers={"User-Agent": "simple-weather/1.0"}
    )
    with urllib.request.urlopen(req, timeout=10) as r:
        return json.load(r)

def load_geocode_cache() -> dict:
    if GEOCODE_CACHE.exists():
        try:
            return json.loads(GEOCODE_CACHE.read_text())
        except Exception:
            return {}
    return {}

def save_geocode_cache(cache: dict) -> None:
    GEOCODE_CACHE.write_text(json.dumps(cache, indent=2, sort_keys=True))

def geocode(place: str) -> dict:
    cache = load_geocode_cache()
    key = place.strip().lower()
    if key in cache:
        return cache[key]

    params = urllib.parse.urlencode({
        "name": place,
        "count": 1,
        "language": "en",
        "format": "json",
    })
    url = f"https://geocoding-api.open-meteo.com/v1/search?{params}"
    data = fetch_json(url)

    results = data.get("results") or []
    if not results:
        raise SystemExit(f"No location found for: {place}")

    r = results[0]
    result = {
        "name": r.get("name"),
        "admin1": r.get("admin1"),
        "country": r.get("country"),
        "latitude": r["latitude"],
        "longitude": r["longitude"],
        "timezone": r["timezone"],
    }
    cache[key] = result
    save_geocode_cache(cache)
    return result

def forecast(lat: float, lon: float, timezone: str) -> dict:
    params = urllib.parse.urlencode({
        "latitude": lat,
        "longitude": lon,
        "timezone": timezone,
        "current": ",".join([
            "temperature_2m",
            "apparent_temperature",
            "weather_code",
            "wind_speed_10m",
            "is_day",
        ]),
        "daily": ",".join([
            "temperature_2m_max",
            "temperature_2m_min",
            "precipitation_probability_max",
        ]),
        "forecast_days": 1,
    })
    url = f"https://api.open-meteo.com/v1/forecast?{params}"
    return fetch_json(url)

def fmt_place(loc: dict) -> str:
    parts = [loc.get("name"), loc.get("admin1"), loc.get("country")]
    return ", ".join(x for x in parts if x)

def main():
    place = " ".join(sys.argv[1:]).strip() or "Madrid, Spain"

    loc = geocode(place)
    data = forecast(loc["latitude"], loc["longitude"], loc["timezone"])

    current = data["current"]
    daily = data["daily"]

    desc = WEATHER_CODES.get(current["weather_code"], f"weather code {current['weather_code']}")
    temp = current["temperature_2m"]
    feels = current["apparent_temperature"]
    wind = current["wind_speed_10m"]

    tmax = daily["temperature_2m_max"][0]
    tmin = daily["temperature_2m_min"][0]
    precip = daily["precipitation_probability_max"][0]

    print(fmt_place(loc))
    print(f"Now: {temp}°C, feels like {feels}°C, {desc}, wind {wind} km/h")
    print(f"Today: high {tmax}°C, low {tmin}°C, max precip probability {precip}%")
    print(f"Observed at: {current['time']} ({loc['timezone']})")

if __name__ == "__main__":
    main()
```
