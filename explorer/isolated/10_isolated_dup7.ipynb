```
def read_msg(n: int = -1, relative: bool = False, id: str = None) -> dict:
    """Read a message from conversation history."""
    # Minimal stub implementation
    if id:
        return {'id': id, 'content': ''}
    return type('Message', (), {'id': 'msg_123'})()

def _get_deps(code: str) -> list:
    """Get dependencies from code."""
    return []

def _get_nb(code: str, deps: list) -> dict:
    """Create notebook dict from code and dependencies."""
    return {'cells': []}
```

```
def _isolated_nb(id: str='') -> dict:
    "Get dialog notebook dict with source code from message and dependencies."
    if not id: id = read_msg(n=-1, relative=True).id
    target_code = read_msg(0, id=id)['content']
    deps = _get_deps(target_code)
    return _get_nb(target_code, deps)
```
